## 1. Installer `uv`

### 1.1. Pourquoi `uv` plutôt que `venv` + `pip` ?

`uv` est un outil moderne, rapide, qui gère à la fois la version de Python, l'environnement virtuel et les dépendances d'un projet. Il remplace, dans ce cours :

- `python -m venv` (création manuelle d'environnement) ;
- `pip install` (installation non tracée des paquets) ;
- des notes éparses du type « chez moi ça marche ».

**Règle de ce cours**

On ne crée jamais un environnement avec `python -m venv` et on n'installe jamais de paquet avec `pip install` directement. Toutes les commandes passent par `uv`.

### 1.2. Installation

Sur macOS ou Linux, dans un terminal :

```bash

```

Sur Windows, dans PowerShell :

```powershell

```

Comme pour tout script d'installation téléchargé depuis Internet, vérifiez la source officielle (`astral.sh/uv`) avant d'exécuter la commande. Sur un poste institutionnel, suivez la procédure validée par l'administrateur réseau.

Fermez puis rouvrez le terminal, puis vérifiez dans le terminal :

## 4. Premier contact avec des données tabulaires

Créons un tout petit jeu de données local - quelques annonces immobilières, dans l'esprit de l'exemple fil rouge du cours (la section « Cycle de vie d'un projet de Machine Learning ») - pour manipuler `pandas` sans dépendre d'Internet.

````python
import pandas as pd

annonces = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "quartier": ["Maarif", "Hay Riad", "Gueliz", "Maarif", "Agdal", "Hay Riad"],
    "surface_m2": [65, 90, 120, 48, 75, 110],
    "nb_chambres": [2, 3, 4, 1, 2, 3],
    "prix_mad": [850_000, 1_450_000, 1_900_000, 520_000, 980_000, 1_680_000],
})

annonces.to_csv("data/raw/annonces.csv", index=False)
print(annonces)
````

### 4.1. Explorer un `DataFrame`

````python
df = pd.read_csv("data/raw/annonces.csv")

print(df.head())        # les premières lignes
print(df.shape)          # (nombre de lignes, nombre de colonnes)
print(df.dtypes)         # le type de chaque colonne
print(df.describe())     # statistiques descriptives des colonnes numériques
````

### 4.2. Sélectionner, filtrer, agréger

````python
# sélectionner une colonne
print(df["prix_mad"])

# sélectionner plusieurs colonnes
print(df[["quartier", "prix_mad"]])

# filtrer des lignes
grands_biens = df[df["surface_m2"] > 80]
print(grands_biens)

# prix moyen par quartier
print(df.groupby("quartier")["prix_mad"].mean().round(0))

# créer une nouvelle colonne
df["prix_m2"] = (df["prix_mad"] / df["surface_m2"]).round(0)
print(df.sort_values("prix_m2", ascending=False))
````

**Remarque**

Ce petit jeu de données est directement lisible par des humains - c'est volontaire. Le chapitre 3 (Lab 2) travaillera sur des données plus grandes et volontairement imparfaites (valeurs manquantes, doublons, incohérences).

### 4.3. Un premier graphique

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["surface_m2"], df["prix_mad"], s=60)
ax.set_xlabel("Surface (m²)")
ax.set_ylabel("Prix (MAD)")
ax.set_title("Prix des annonces selon la surface")
fig.tight_layout()
fig.savefig("outputs/prix_vs_surface.png", dpi=150)
plt.show()
````

Même avec six observations, la relation entre surface et prix commence déjà à se dessiner à l'œil nu - c'est précisément ce type de relation qu'un modèle de régression (chapitre 4 / Lab 3) apprendra à quantifier.

## 5. Récupérer des données réelles via une API publique

### 5.1. Principe

Une lecture de données en ligne suit toujours le même schéma : construire une URL, envoyer une requête HTTP, vérifier le code de réponse, puis interpréter le contenu (souvent du JSON).

````python
import requests

response = requests.get(
    "https://api.github.com/repos/pandas-dev/pandas",
    timeout=20,
)
response.raise_for_status()   # lève une erreur si le code n'est pas 2xx
print(response.json()["full_name"])
````

**Toujours fixer un `timeout`**

Sans `timeout`, un programme peut rester bloqué indéfiniment si le serveur ne répond jamais. `raise_for_status()` transforme un code d'erreur HTTP (404, 500...) en exception Python explicite, plutôt que de laisser passer silencieusement une réponse invalide.

### 5.2. Exemple : l'API météo Open-Meteo (sans clé)

````python
import pandas as pd
import requests

params = {
    "latitude": 33.5731,       # Casablanca
    "longitude": -7.5898,
    "hourly": "temperature_2m",
    "timezone": "Africa/Casablanca",
    "forecast_days": 2,
}

response = requests.get(
    "https://api.open-meteo.com/v1/forecast",
    params=params,
    timeout=30,
)
response.raise_for_status()
payload = response.json()

meteo = pd.DataFrame(payload["hourly"])
meteo["time"] = pd.to_datetime(meteo["time"])
print(meteo.head())
````

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(meteo["time"], meteo["temperature_2m"])
ax.set_xlabel("Heure")
ax.set_ylabel("Température (°C)")
ax.set_title("Prévisions horaires - Casablanca")
fig.autofmt_xdate()
fig.tight_layout()
fig.savefig("outputs/meteo_casablanca.png", dpi=150)
plt.show()
````

**Une requête peut échouer**

URL incorrecte, absence de connexion, délai dépassé, service temporairement indisponible : une requête réseau n'est jamais garantie. Dans un vrai projet, on entoure ce type d'appel d'une gestion d'erreurs (`try`/`except`) et on sauvegarde une copie brute de la réponse pour pouvoir rejouer l'analyse hors ligne.

## 6. Vérifier que `scikit-learn` fonctionne : le filtre anti-spam du chapitre 1 

Le chapitre 1 (la section « Paradigme Data-Driven ») présente l'idée du filtre anti-spam comme premier exemple du paradigme Data-Driven. Reproduisons-le concrètement, avec un tout petit jeu d'e-mails fait main.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

emails_entrainement = [
    "Gagnez de l'argent facilement, cliquez ici maintenant",
    "Offre urgente : gagnez un iPhone gratuit aujourd'hui",
    "Réunion projet reportée à demain 10h",
    "Voici le compte-rendu de la réunion d'hier",
    "Urgent : votre compte a gagné un prix, réclamez-le",
    "Peux-tu relire mon rapport avant vendredi ?",
]

y_entrainement = ["spam", "spam", "non-spam", "non-spam", "spam", "non-spam"]

vectoriseur = TfidfVectorizer()
X_entrainement = vectoriseur.fit_transform(emails_entrainement)

modele = LogisticRegression()
modele.fit(X_entrainement, y_entrainement)

nouveaux_emails = [
    "Cliquez ici pour gagner un cadeau gratuit",
    "Peux-tu m'envoyer le fichier avant la réunion ?",
]

X_nouveaux = vectoriseur.transform(nouveaux_emails)
predictions = modele.predict(X_nouveaux)

for email, prediction in zip(nouveaux_emails, predictions):
    print(f"[{prediction}] {email}")
````

Si ce code s'exécute et affiche une prédiction `spam` ou `non-spam` pour chaque e-mail, votre environnement est prêt pour la suite du cours : `scikit-learn`, comme discuté au chapitre 1, a bien **appris** une règle de décision à partir des six exemples fournis, sans qu'aucune règle `SI ... ALORS ...` n'ait été écrite à la main.

**Pourquoi si peu d'exemples ne suffisent pas en pratique**

Six exemples d'entraînement, c'est très peu : le modèle risque de mal généraliser à des e-mails très différents de ceux vus à l'entraînement (voir la notion de généralisation, chapitre 2 / Lab 1). Ce mini-exemple sert uniquement à vérifier que l'environnement fonctionne de bout en bout.
